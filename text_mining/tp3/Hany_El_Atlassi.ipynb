```python
# import libraries

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score
```

```python
documents = ["This ltitle kitty came to play when I was eating at a restaurant.",
            "Merley has the bets squooshy kitten belly.",
            "Google Translate app is incredible.",
            "If you open 100 tab in google you get smiley face.",
            "Best cat photo I've ever taken.",
            "Climbing ninja cat.",
            "Impressed with google map feedback.",
            "Key promoter extension for Google Chrome."
             ]
```

```python
vectorizer = TfidfVectorizer(stop_words='english')
```

```python
x = vectorizer.fit_transform(documents)
```

```python
print(vectorizer.vocabulary_)
```

```python
print(vectorizer.get_feature_names_out())
```

```python
true_k = 2
model = KMeans(n_clusters=true_k, init="k-means++", max_iter=100, n_init=1)
model.fit(x)
```

```python
print(model.labels_)
```

```python
print("Prediction")
y = vectorizer.transform(["chrome browser to open"])
prediction = model.predict(y)
print(prediction)
```

```python
y = vectorizer.transform(["My cat is hungry."])
prediction = model.predict(y)
print(prediction)
```

```python
print("Top terms per cluster:")
order_centroids = model.cluster_centers_.argsort()[:,::-1]
terms = vectorizer.get_feature_names_out()
for i in range(true_k):
    print("Cluster %d: " %i),
    for ind in order_centroids[i,:10]:
        print(" %s "% terms[ind])
    print
```

```python
import pandas as pd

results = []

# we only have 8 documents, can't have k > 8.

for k in range(3, 9):
    model = KMeans(
        n_clusters=k,
        init="k-means++",
        max_iter=100,
        n_init=1,
        random_state=42
    )
    model.fit(x)

    order_centroids = model.cluster_centers_.argsort()[:, ::-1]

    for cluster_id in range(k):
        top_terms = [terms[index] for index in order_centroids[cluster_id, :10]]
        results.append({
            "k": k,
            "Cluster": cluster_id,
            "Top terms": ", ".join(top_terms)
        })

results_table = pd.DataFrame(results)
results_table
```
