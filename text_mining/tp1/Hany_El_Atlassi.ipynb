### 1.2 Comptage de mots dans différents documents

```python
from nltk.corpus import brown
news_text = brown.words(categories="news")
fdist = nltk.FreqDist([w.lower() for w in news_text])
modals = ['can', 'could', 'may', 'might','must', 'will']
for m in modals:
    print(m + ":", fdist[m])
```

```python
from nltk.corpus import brown
news_text = brown.words(categories="science_fiction")
fdist = nltk.FreqDist([w.lower() for w in news_text])
modals = ['who', 'what', 'where', 'why', 'when', 'whom', 'whose', 'which']
for m in modals:
    print(m + ":", fdist[m])
```

```python
cfd = nltk.ConditionalFreqDist((genre, word) for genre in brown.categories() for word in brown.words(categories=genre))
genres = ['news', 'religion', 'hobbies', 'science_fiction', 'romance', 'humor']
cfd.tabulate(conditions = genres, samples = modals)
```

### 1.3 Les stopwords

```python
def unusual_words(text):
    text_vocab = set(w.lower() for w in text if w.isalpha())
    english_vocab = set(w.lower() for w in nltk.corpus.words.words())
    unusual = text_vocab.difference(english_vocab)
    return sorted(unusual)
```

```python
unusual_words(nltk.corpus.gutenberg.words('austen-sense.txt'))
```

```python
unusual_words(nltk.corpus.nps_chat.words())
```

```python
from nltk.corpus import stopwords
stopwords.words('english')
```

```python
def content_fraction(text):
    stopwords = nltk.corpus.stopwords.words('english')
    content = [w for w in text if w.lower() not in stopwords]
    return len(content)/len(text)
```

```python
content_fraction(nltk.corpus.reuters.words())
```

### 2.2 Stemming et lemmatisation

```python
raw = """DENNIS: Listen, strange women lying in ponds distirbuting swords is no basis for a system of government. Supreme executive power derives from a mandate from the masses, not from some farcical aquatic ceremony."""
tokens = nltk.word_tokenize(raw)
```

```python
porter = nltk.PorterStemmer()
[porter.stem(t) for t in tokens]
```

```python
lancaster = nltk.LancasterStemmer()
[lancaster.stem(t) for t in tokens]
```

```python
wnl = nltk.WordNetLemmatizer()
[wnl.lemmatize(t) for t in tokens]
```
